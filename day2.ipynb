##### Data Ingestion:

```python
from pyspark.sql import SparkSession
```

```python
spark = (
    SparkSession.builder
    .appName("data_ingestion_app")
    .master("local[*]")
    .getOrCreate()
)
```

```python
# Read csv file:
df_csv = (
    spark.read
    .option("header", True)
    .option("inferschema", True)
    .csv("./data/products.csv")
)
```

```python
df_csv.show()
```

```python
# Read from json file:

df_json = (
    spark.read
    .option("header", False)
    .option("multiline", True)
    .option("inferschema", False)
    .json("./data/suppliers.json")
)
```

```python
df_json.show()
# df_json.printSchema()
```

```python
df_parquet = (
    spark.read
    .parquet("./data/customers.parquet")
)
```

```python
# df_parquet.show()
df_parquet.printSchema()
```

```python
df_txt = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option('sep', "|")
    .csv("./data/orders.txt")
)
```

```python
df_txt.show()
```

```python
df_orc = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .orc("./data/payments.orc")
)
```

```python
# df_orc.show()
df_orc.printSchema()
```

```python
import requests
import json
```

```python
url = "https://api.open-meteo.com/v1/forecast?latitude=17.38&longitude=78.48&current_weather=true"

data= requests.get(url=url).json()

df_api = spark.read.json(spark.sparkContext.parallelize([json.dumps(data["current_weather"])]))
```
