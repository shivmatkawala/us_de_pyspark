```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, DateType
```

```python
spark = (
    SparkSession.builder
    .appName("hr_app")
    .master("local[*]")
    .getOrCreate()
)
```

```python
hr_employee_schema = StructType([
    StructField("employee_id", StringType(), nullable=False),
    StructField("full_name", StringType(), nullable=True),
    StructField("email", StringType(), nullable=True),
    StructField("department", StringType(), nullable=True),
    StructField("job_title", StringType(), nullable=True),
    StructField("hire_date", StringType(), nullable=True),
    StructField("salary", StringType(), nullable=True),
    StructField("city", StringType(), nullable=True),
    StructField("gender", StringType(), nullable=True),
    StructField("phone", StringType(), nullable=True)
])
```

```python
df = (
    spark.read
    .option("header", True)
    .schema(hr_employee_schema)
    .csv("./data/hr_employees_dirty.csv")
)
```
